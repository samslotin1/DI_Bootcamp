```python
import numpy as np
from scipy import stats

productivity_before = np.random.normal(loc=50, scale=10, size=30)

productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)

mean_before = np.mean(productivity_before)
mean_after = np.mean(productivity_after)

print(mean_before)
print(mean_after)

test = stats.ttest_rel(productivity_before, productivity_after)

print(test)
```

The paired t-test produced a p-value of 0.0000163, which is less than 0.05. Therefore, we reject the null hypothesis and conclude that the training program significantly improved employee productivity.

```python
import numpy as np

dates = np.arange(
    np.datetime64('2023-01-01'),
    np.datetime64('2024-01-01'),
    dtype='datetime64[D]'
)

jan_mar = dates[
    (dates >= np.datetime64('2023-01-01')) &
    (dates <= np.datetime64('2023-03-31'))
]

apr_jun = dates[
    (dates >= np.datetime64('2023-04-01')) &
    (dates <= np.datetime64('2023-06-30'))
]

jul_sep = dates[
    (dates >= np.datetime64('2023-07-01')) &
    (dates <= np.datetime64('2023-09-30'))
]

oct_dec = dates[
    (dates >= np.datetime64('2023-10-01')) &
    (dates <= np.datetime64('2023-12-31'))
]

print(jan_mar)
print(apr_jun)
print(jul_sep)
print(oct_dec)
```
