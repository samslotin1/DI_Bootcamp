```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv(
    "ex2data1.txt",
    header=None,
    names=["Exam1", "Exam2", "Admitted"]
)

# Split into groups
admitted = df[df["Admitted"] == 1]
not_admitted = df[df["Admitted"] == 0]

# Scatter plot
plt.figure(figsize=(8, 6))

plt.scatter(
    admitted["Exam1"],
    admitted["Exam2"],
    c="blue",
    label="Admitted"
)

plt.scatter(
    not_admitted["Exam1"],
    not_admitted["Exam2"],
    c="red",
    label="Not Admitted"
)

plt.xlabel("Exam 1 Score")
plt.ylabel("Exam 2 Score")
plt.title("Student Admission by Exam Scores")
plt.legend()
plt.show()
```

The scatter plot shows that students with higher scores on both exams are more likely to be admitted. Students with lower exam scores are generally not admitted. The data appears to show a pattern that could be separated by a logistic regression model.

```python
from sklearn.linear_model import LogisticRegression

X = df[["Exam1", "Exam2"]]

y = df["Admitted"]

model = LogisticRegression()

model.fit(X, y)
```

```python
from sklearn.metrics import accuracy_score

y_pred = model.predict(X)

accuracy = accuracy_score(y, y_pred)

print("Accuracy:", accuracy)
```

The logistic regression model achieved an accuracy of 89%. This means the model correctly predicted whether a student would be admitted or not admitted 89% of the time. The result suggests that Exam 1 and Exam 2 scores are strong indicators of admission status and that the model performs well on this dataset.
