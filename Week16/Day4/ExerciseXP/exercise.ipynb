## Exercise 1: Rewrite and Optimize a Vague Prompt
**Scenario**: Manager asks for a LinkedIn post about FlowNest with the vague prompt `"Write something about productivity tips."`

### Tasks
- Identify 3 critical issues with the prompt.
- Rewrite it using: role prompting, clear instructions, and format/tone/length constraints.
- Add a final line to make it reusable for future social posts.

```python
# Completed: list three critical issues with the vague prompt
critical_issues = [
    "It does not define the target audience, so the post may not speak to the right users, such as busy professionals, students, or remote teams.",
    "It gives no brand or product context for FlowNest, so the output may become generic productivity advice instead of promoting the app.",
    "It does not specify LinkedIn format, tone, length, or engagement goal, so the post may be too vague, too long, or missing a call to action.",
]
critical_issues
```

```python
# Completed: rewrite the prompt using role, instructions, and constraints
rewritten_prompt = """Act as a B2B social media copywriter for FlowNest, a focus app that helps busy professionals block distractions and plan deep-work sessions.

Write a LinkedIn post promoting FlowNest with 3 actionable productivity tips. Connect each tip to a clear benefit of the app.

Use a confident, helpful, and friendly tone. Keep the post under 180 words. Include a short hook, 3 bullet points, and a final call to action inviting readers to try FlowNest.

Use this same structure whenever we promote productivity tools on social media."""
rewritten_prompt
```

## Exercise 2: Multi-Part Prompt for Quiz Generation
**Scenario**: Create an educational prompt for 7th graders about volcanic eruptions.

### Tasks
- Ask for: 2-bullet summary + 3 MCQs (1 correct, 2 distractors).
- Require friendly tone, simple vocab (11?13 y/o), and Google Slides-ready output.
- Explain 3 ways your prompt improves on: `Make a quiz for kids about this article.`

```python
# Completed: write the full multi-part prompt
quiz_prompt = """Act as a 7th-grade science teacher creating Google Slides content from the article pasted below.

Use only the information in the article. Write for students aged 11-13 using simple vocabulary and a friendly, encouraging tone.

Create the output in this Google Slides-ready format:

Slide 1: Title
- Give a clear title about volcanic eruptions.

Slide 2: Quick Summary
- Summarize the article in exactly 2 simple bullet points.

Slides 3-5: Quiz Questions
- Write 3 multiple-choice questions.
- Each question must have 3 answer choices: A, B, and C.
- Include exactly 1 correct answer and 2 believable distractors.
- Mark the correct answer after each question using: Correct answer: [letter].

Article:
[Paste article here]"""
quiz_prompt
```

```python
# Completed: explain three improvements over the vague prompt
improvements = [
    "It defines the audience as 7th-grade students aged 11-13, which keeps vocabulary and difficulty age-appropriate.",
    "It specifies the exact output structure: 2 summary bullets, 3 multiple-choice questions, 3 choices per question, and correct-answer labels.",
    "It requires the model to use only the provided article, reducing irrelevant content and unsupported facts.",
]
improvements
```

```python
# Completed: choose a style and write the prompt
chosen_style = "Structured"
prompt_template = """Act as a customer support chatbot for an e-commerce company.

When a customer asks for help, respond in this structure:

1. Acknowledge the issue in one friendly sentence.
2. Ask for any missing required details, such as order number, email address, product name, or delivery date.
3. Provide the next best action in 1-3 clear steps.
4. If the issue involves refunds, damaged items, missing packages, or account access, explain that a human support agent may need to review it.
5. End with a short reassurance.

Use a calm, helpful tone. Do not promise refunds, delivery dates, or policy exceptions unless they are confirmed in the provided company policy."""
key_style_features = [
    "It uses a fixed response structure with numbered sections and clear steps.",
    "It defines escalation rules and limits what the chatbot can promise.",
]
justification = (
    "Structured is the best style because customer support needs consistent, complete, and easy-to-follow answers. "
    "Exploratory would be too open-ended, conversational could miss required details, and purely functional might feel too cold for customers."
)
{
    "chosen_style": chosen_style,
    "prompt_template": prompt_template,
    "key_style_features": key_style_features,
    "justification": justification,
}
```

## Exercise 6: Hallucination Spotting and Mitigation
**Scenario**: Model added an unsupported claim in a climate/marine article summary.

### Tasks
1) Prompt the model to avoid hallucinations using only provided text.
2) Rewrite with verification language.
3) List 2 mitigation strategies.
4) Name 2 domains where hallucinations are risky and why.

```python
# Completed: baseline anti-hallucination prompt
no_hallucination_prompt = """Act as a careful research summarizer.

Summarize the peer-reviewed article pasted below using only information explicitly stated in the text. Do not add outside facts, predictions, statistics, or claims from general knowledge.

If the article does not mention a number, cause, date, species, or conclusion, do not include it. Use short bullets and keep the wording faithful to the article.

Article:
[Paste article here]"""
no_hallucination_prompt
```

```python
# Completed: verification-focused rewrite
verification_prompt = """Summarize the provided article on climate change and marine biodiversity.

Verification rules:
- Only include claims explicitly stated in the article.
- For every statistic or projection, include the sentence or section it comes from.
- If uncertain, write: 'This information is not available in the article.'
- Do not include unsupported extinction percentages, dates, or predictions.
- Before finalizing, check each claim against the article and remove anything that cannot be verified.

Output:
1. Verified summary bullets
2. Key statistics found in the article
3. Information not available in the article"""
verification_prompt
```

```python
# Completed: list mitigation strategies and risky domains
mitigation_strategies = [
    "Restrict the model to the provided source text and explicitly forbid outside facts or unsupported claims.",
    "Require verification language: cite where each claim comes from and state when information is not available.",
]
risky_domains = {
    "domain_1": {
        "name": "Healthcare",
        "why": "A hallucinated diagnosis, dosage, or treatment claim could harm patient safety and lead to dangerous decisions.",
    },
    "domain_2": {
        "name": "Legal",
        "why": "A hallucinated law, case citation, or compliance requirement could mislead legal strategy and create serious liability.",
    },
}
{
    "mitigation_strategies": mitigation_strategies,
    "risky_domains": risky_domains,
}
```
