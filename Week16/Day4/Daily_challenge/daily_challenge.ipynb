# Exercises XP: Advanced Prompting

This Google Colab notebook completes the prompt development workflow for turning a remote access security policy paragraph into employee-facing microlearning content.


## Source Policy Text

> Employees must ensure that all remote access to internal systems is established via the approved secure VPN. Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications.


## Step 1: Prompt That Controls Tone, Format, and Length


```python
source_text = """Employees must ensure that all remote access to internal systems is established via the approved secure VPN. Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications."""

step_1_prompt = """Act as an internal communications specialist.

Rewrite the policy text below into a friendly, clear employee newsletter snippet.

Requirements:
- Use bullet points.
- Paraphrase the meaning without direct quotes.
- Keep the total response under 75 words.
- Keep the message accurate and easy for non-technical employees to understand.

Policy text:
{source_text}
"""

print(step_1_prompt)
```

## Step 1: Generated Response


```python
generated_response = """- Use the approved secure VPN whenever you connect to internal systems remotely.
- Do not use unsecured connections to view company data or sensitive messages.
- Only use devices with endpoint protection before accessing proprietary information."""

print(generated_response)
print("\nWord count:", len(generated_response.replace("-", "").split()))
```

## Step 2: Evaluate the Output


```python
evaluation = {
    "Relevance": "The output sticks closely to the original message about secure remote access.",
    "Clarity": "The language is easier for non-technical staff to understand.",
    "Structure": "The response uses clear bullet points.",
    "Tone": "The tone is friendly and suitable for internal employee communication.",
    "Length": "The response is under 75 words.",
    "Factual Accuracy": "No critical security details were invented or removed.",
}

for criterion, result in evaluation.items():
    print(f"{criterion}: {result}")
```

## Step 3: Detect and Mitigate Hallucinations

If the model adds extra policy terms, tools, punishments, technologies, or examples that do not appear in the input, use a stricter prompt.


```python
hallucination_mitigation_prompt = """Act as an internal communications specialist.

Only paraphrase the content in the input text. Do not add new recommendations, technologies, examples, consequences, tools, or policy terms.

Rewrite the policy text in a friendly and clear tone.

Requirements:
- Use bullet points.
- Stay under 75 words total.
- Keep all security details from the original text.
- If a detail is not in the input, do not include it.

Policy text:
{source_text}
"""

print(hallucination_mitigation_prompt)
```

## Step 4: Paraphrasing Deep Dive for Junior Interns


```python
junior_intern_prompt = """Act as a supportive workplace mentor.

Paraphrase the policy text below for junior interns using plain language and short phrases.

Requirements:
- Use no more than 4 bullet points.
- Avoid corporate, legal, or technical jargon.
- Keep the tone supportive and informative.
- Do not add information that is not in the original text.

Policy text:
{source_text}
"""

junior_intern_output = """- Use the company-approved secure VPN when working remotely.
- Do not connect through unsafe networks.
- Do not use personal devices unless they are protected.
- Keep company data and private messages safe."""

print("Prompt:\n")
print(junior_intern_prompt)
print("\nSample output:\n")
print(junior_intern_output)
```

## Step 5: Quote Extraction Variant


```python
quote_extraction_prompt = """Read the policy text below and extract one direct quote that best captures the core security policy.

Requirements:
- Choose only one quote.
- Do not paraphrase.
- Do not add explanation inside the quote.
- Select the shortest quote that still captures the main rule.

Policy text:
{source_text}
"""

best_quote = "all remote access to internal systems is established via the approved secure VPN"

print("Prompt:\n")
print(quote_extraction_prompt)
print("\nBest quote:")
print(f'"{best_quote}"')
```

## Step 5: When Quoting Is Useful or Risky


```python
quote_reflection = {
    "When quoting is more appropriate": (
        "Quoting is useful in formal policy reminders, compliance training, legal notices, "
        "or security announcements where exact wording matters."
    ),
    "When quoting may pose a risk": (
        "Quoting may be risky when the original wording is too technical, too formal, "
        "or easy for employees to misunderstand. It can also confuse readers if they need plain-language guidance."
    ),
}

for question, answer in quote_reflection.items():
    print(f"{question}: {answer}")
```
