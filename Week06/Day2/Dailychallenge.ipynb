```python
import numpy as np
import pandas as pd

data = np.random.uniform(-5, 35, size=(10, 12))
print(data)

cities = ["Jerusalem", "Savannah", "Santa Cruz", "Charlotte", "Raleigh", "Tel Aviv", "Dubai", "Atlanta", "Asheville", "Boone"]
months = ["January",
    "February",
    "March",
    "April",
    "May",
    "June",
    "July",
    "August",
    "September",
    "October",
    "November",
    "December"]

df = pd.DataFrame(data, index=cities, columns=months)
print(df)
```

```python
annual_avg = df.mean(axis=1)
print(annual_avg)
```

```python
highest = annual_avg.idxmax()
lowest = annual_avg.idxmin()
print(highest)
print(lowest)
print(annual_avg[highest])
print(annual_avg[lowest])
```

```python
import matplotlib.pyplot as plt

plt.bar(cities, annual_avg)
plt.xticks(rotation=45, ha='right')
plt.title("Annual Average Temperature by City")
plt.show()
```
